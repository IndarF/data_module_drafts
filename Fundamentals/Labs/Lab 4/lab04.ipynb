### Histogram: Grams of Sugar

Below is a Histogram created from Frozen_Desserts.csv, <b>‘Grams_of_Sugar’</b> column. 

A Histogram can be created for a column in the .csv data frame by importing the matplotlib library and creating a call to the column by name.

`Frozen_desserts.csv` 

<img src="https://github.com/IndarF/data_module_drafts/blob/main/Fundamentals/Labs/Lab%203/frozen_desserts.jpg?raw=true" width="800"/>

Here is the code to plot the histogram:

```python
import pandas as pd
import matplotlib.pyplot as plt

data_frame = pd.read_csv("Frozen_desserts.csv", sep=",")
grams_of_sugar = data_frame['Grams of Sugar']
grams_of_sugar.hist()
plt.title('Grams of Sugar')
plt.ylabel('Frequency')
plt.xlabel('Sugar(In Grams)')
plt.show()
```

Here is the expected output:

When we talk about estimating the typical value of a dataset, we are trying to find a number, or range of numbers, that best represents what is common in the data. 

Often, the typical value is somewhere near the center of the distribution. When we use a histogram to estimate the typical value, we often look for the interval (or bin(s)) where the data is most concentrated. Because a histogram groups data into ranges, the typical value is usually described as a range rather than a single exact number. 

While there are formal calculations to find measures of center (like the mean and median), it’s important to practice making visual estimates based on graphs.
