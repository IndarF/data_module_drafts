## Part 2: Measures of Central Tendency

Now that we are a little familiar with python, we can look at what other tools python has. In Data Science, we like to look at data sets and measure their `central tendancy`. Central tendency tends to be values that represent a typical, or central, entry of a data set. The most commonly used measures of central tendency are:

- Mean
- Median
- Mode

We will be analyzing each neasure in detail.

### Median

A median is a value that lies in the middle of the data when the data set is in numerical order. This measures the center of an ordered list by divings it into two equal parts. Lets find the median of an arbitrary data set.

```python
import numpy as np    # Importing library with tools to find central tendency. Will learn about this later
from scipy import stats as st   

data_set = [32, 23, 1, 4, 9]    # Feel free to replace the numbers inside of the brackets
np.median(data_set)
```

### Mean

A mean is the sum of all the data entries divided by the number of entries. Another word for mean would be average, which is commonnly used. Like before, lets use an arbitrary data set and find its mean. 

```python
data_set = [3, 34, 22, 11, 36, 3, 7, 9, 4]    # Use any numbers you would like
np.mean(data_set)
```

### Mode

The mode is the data entry that occurs with the greatest frequency. If no entry of data is repeated, then there is no mode. Lets use an arbitraty data set to find its mode.

```python
data_set = [1, 1, 23, 4, 56, 32, 1, 3, 4, 3]     # Change the numbers as you see fit
st.mode(data_set)
```

These are some ways to measure the central tendency of data sets in python. Feel free to compare the results of these measures with the same data set. There are pros and cons for using each measure.
